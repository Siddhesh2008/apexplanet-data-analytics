```python
# Import pandas
import pandas as pd

# Write an SQL query to retrieve 10 transactions
query = """
SELECT
    transaction_id,
    user_name,
    country,
    purchase_amount
FROM public.ecommerce_transactions
LIMIT 10;
"""

# Execute the SQL query and store the result in a DataFrame
df = pd.read_sql(query, engine)

# Display the retrieved data
df
```

##### Business Question 1: Which are the top 5 product categories by revenue?

```python
# SQL query to find the top 5 categories by revenue
query = """
SELECT
    product_category,
    SUM(purchase_amount) AS total_revenue
FROM public.ecommerce_transactions
GROUP BY product_category
ORDER BY total_revenue DESC
LIMIT 5;
"""

# Execute the SQL query and store results in a DataFrame
top_categories = pd.read_sql(query, engine)

# Display the results
top_categories
```

##### Business Question 2: What is the monthly sales trend?

```python
# SQL query to calculate monthly revenue
query = """
SELECT
    DATE_TRUNC('month', transaction_date) AS month,
    SUM(purchase_amount) AS monthly_revenue
FROM public.ecommerce_transactions
GROUP BY DATE_TRUNC('month', transaction_date)
ORDER BY month;
"""

# Execute the query and store the results in a DataFrame
monthly_sales = pd.read_sql(query, engine)

# Display the results
monthly_sales
```

##### Business Question 3: Which payment methods generate the highest revenue?

```python
# Calculate total revenue for each payment method
query = """
SELECT
    payment_method,
    COUNT(*) AS total_transactions,
    SUM(purchase_amount) AS total_revenue,
    AVG(purchase_amount) AS average_purchase
FROM public.ecommerce_transactions
GROUP BY payment_method
ORDER BY total_revenue DESC;
"""

# Execute the SQL query using Python
payment_analysis = pd.read_sql(query, engine)

# Display the results
payment_analysis
```

##### Business Question 4: Which countries generate the highest revenue?

```python
# Calculate revenue generated by each country
query = """
SELECT
    country,
    COUNT(*) AS total_transactions,
    SUM(purchase_amount) AS total_revenue,
    AVG(purchase_amount) AS average_purchase
FROM public.ecommerce_transactions
GROUP BY country
ORDER BY total_revenue DESC;
"""

# Execute the query and store the results
country_analysis = pd.read_sql(query, engine)

# Display the results
country_analysis
```

##### Business Question 5: Who are the highest-spending customers?

```python
# Find the 10 highest-spending user names
query = """
SELECT
    user_name,
    COUNT(*) AS total_transactions,
    SUM(purchase_amount) AS total_spent,
    AVG(purchase_amount) AS average_purchase
FROM public.ecommerce_transactions
GROUP BY user_name
ORDER BY total_spent DESC
LIMIT 10;
"""

top_spenders = pd.read_sql(query, engine)

top_spenders
```

##### Question 6: Which age groups generate the most revenue?

```python
# Group transactions by age range and calculate spending
query = """
SELECT
    CASE
        WHEN age BETWEEN 18 AND 25 THEN '18-25'
        WHEN age BETWEEN 26 AND 35 THEN '26-35'
        WHEN age BETWEEN 36 AND 45 THEN '36-45'
        WHEN age BETWEEN 46 AND 55 THEN '46-55'
        ELSE '56-70'
    END AS age_group,
    COUNT(*) AS total_transactions,
    SUM(purchase_amount) AS total_revenue,
    AVG(purchase_amount) AS average_purchase
FROM public.ecommerce_transactions
GROUP BY age_group
ORDER BY total_revenue DESC;
"""

age_analysis = pd.read_sql(query, engine)
age_analysis
```

##### Question 7: Which product categories have the most transactions?

```python
# Count transactions in each product category
query = """
SELECT
    product_category,
    COUNT(*) AS total_transactions,
    SUM(purchase_amount) AS total_revenue
FROM public.ecommerce_transactions
GROUP BY product_category
ORDER BY total_transactions DESC;
"""

category_transactions = pd.read_sql(query, engine)
category_transactions
```

##### Question 8: Which countries have the highest average purchase amount?

```python
# Compare average purchase amounts across countries
query = """
SELECT
    country,
    AVG(purchase_amount) AS average_purchase,
    COUNT(*) AS total_transactions
FROM public.ecommerce_transactions
GROUP BY country
ORDER BY average_purchase DESC
LIMIT 5;
"""

country_average = pd.read_sql(query, engine)
country_average
```

##### Question 9: How does transaction volume change each month?

```python
# Count transactions for each month
query = """
SELECT
    DATE_TRUNC('month', transaction_date) AS month,
    COUNT(*) AS total_transactions
FROM public.ecommerce_transactions
GROUP BY DATE_TRUNC('month', transaction_date)
ORDER BY month;
"""

monthly_transactions = pd.read_sql(query, engine)
monthly_transactions
```

##### Question 10: Which payment methods have the highest average purchase amount?

```python
# Compare average purchase amounts by payment method
query = """
SELECT
    payment_method,
    AVG(purchase_amount) AS average_purchase,
    COUNT(*) AS total_transactions
FROM public.ecommerce_transactions
GROUP BY payment_method
ORDER BY average_purchase DESC;
"""

payment_average = pd.read_sql(query, engine)
payment_average
```
